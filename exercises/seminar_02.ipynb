# Data Schema Information
- The sample data has 6 columns with 500 entries
- Some values are missing in columns Quantity and Price
- Some values have wrong data type like the Product, Category, OrderDate





```python
test.info()
```

```python
test['Product'] = test['Product'].astype('string')
```

```python
test['Category'] = test['Category'].astype('string')
```

```python
test['OrderDate'] = pd.to_datetime(test['OrderDate'], format='mixed')
```
