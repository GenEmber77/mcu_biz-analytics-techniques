```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
trees = pd.read_csv("trees_seminar.csv")
trees.isnull().sum()
```

```python
trees.describe()
```

```python
trees.replace(['', ' ', 'NaN', 'null', 'NULL', 'None', 'NA'], pd.NA, inplace=True)
trees.isnull().sum()
```

```python
trees.columns
```

```python
trees['Trunk_Height'] = trees['Trunk_Height'].fillna(trees['Trunk_Height'].median())
trees['Leaf_Width'] = trees['Leaf_Width'].fillna(trees['Leaf_Width'].median())
trees.isnull().sum()
```

```python
def draw_boxplot():
    plt.figure(figsize=(16, 8))

    # Trunk_Height boxplot
    # patch_artist=True allows you to fill the inside of boxplots with color
    plt.subplot(2, 2, 1)
    plt.boxplot(trees['Trunk_Height'], patch_artist=True)
    plt.title('Boxplot of Trunk Height')
    plt.ylabel('Trunk Height (meters)')

    # Trunk_Diameter boxplot
    plt.subplot(2, 2, 2)
    plt.boxplot(trees['Trunk_Diameter'], patch_artist=True)
    plt.title('Boxplot of Trunk Diameter')
    plt.ylabel('Trunk Diameter (meters)')

    # Leaf_Length boxplot
    plt.subplot(2, 2, 3)
    plt.boxplot(trees['Leaf_Length'], patch_artist=True)
    plt.title('Boxplot of Leaf Length')
    plt.ylabel('Leaf Length (cm)')

    # Leaf_Width boxplot
    plt.subplot(2, 2, 4)
    plt.boxplot(trees['Leaf_Width'], patch_artist=True)
    plt.title('Boxplot of Leaf Width')
    plt.ylabel('Leaf Width (cm)')

    trees.describe()

draw_boxplot()
```

```python
num_cols = trees.select_dtypes(include='number').columns
num_cols
```

```python
for col in num_cols:
    print("")
    print(f"Column Name:{col}")
    median_val = trees[col].median() 
    print(f"median: {median_val}")
    
    #caluclate 25th percentile of leaf length
    percentile25 = trees[col].quantile(0.25)
    print(f"percentile25: {percentile25}")

    #caluclate 75th percentile of leaf length
    percentile75 = trees[col].quantile(0.75)
    print(f"percentile75: {percentile75}")
    
    #Calculate interquartile range
    iqr = percentile75 - percentile25
    print(f"iqr: {iqr}")

    #calcualte upper and lower thresholds for outliers
    upper_limit = percentile75 + 1.5 *iqr
    lower_limit = percentile25 - 1.5 * iqr
    
    df_lower = trees[trees[col] < lower_limit]
    outliers_lower = df_lower[col].tolist()
    trees[col].replace(outliers_lower, median_val, inplace=True)
    print(f"outliers_lower: {outliers_lower}")
    
    df_higher = trees[trees[col] > upper_limit]
    outliers_higher = df_higher[col].tolist()
    trees[col].replace(outliers_higher, median_val, inplace=True)
    print(f"outliers_higher: {outliers_higher}")

    
```

```python
draw_boxplot()
```

```python
# Create boxplots to identify outliers
plt.figure(figsize=(16, 8))

# Trunk_Height boxplot
plt.subplot(2, 2, 1)
plt.hist(trees['Trunk_Height'], bins=20, edgecolor='black')
plt.title('Histogram of Trunk Height')
plt.ylabel('Trunk Height (meters)')

# Trunk_Diameter boxplot
plt.subplot(2, 2, 2)
plt.hist(trees['Trunk_Diameter'], bins=20, edgecolor='black')
plt.title('Histogram of Trunk Diameter')
plt.ylabel('Trunk Diameter (meters)')

# Leaf_Length boxplot
plt.subplot(2, 2, 3)
plt.hist(trees['Leaf_Length'], bins=20, edgecolor='black')
plt.title('Histogram of Leaf Length')
plt.ylabel('Leaf Length (cm)')

# Leaf_Width boxplot
plt.subplot(2, 2, 4)
plt.hist(trees['Leaf_Width'], bins=20, edgecolor='black')
plt.title('Histogram of Leaf Width')
plt.ylabel('Leaf Width (cm)')
```
