```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
```

```python
trees = pd.read_csv('trees_seminar.csv')
trees.head()
```

```python
trees.replace(['', ' ', 'NaN', 'null', 'NULL', 'None', 'NA'], pd.NA, inplace=True)

trees.isnull().sum()
```

```python
trees['Trunk_Height'] = trees['Trunk_Height'].fillna(trees['Trunk_Height'].median())
trees.isnull().sum()
trees['Leaf_Width'] = trees['Leaf_Width'].fillna(trees['Leaf_Width'].median())
trees.isnull().sum()
```

```python
plt.figure(figsize=(16, 8))


plt.subplot(2, 2, 1)
plt.boxplot(trees['Trunk_Height'], patch_artist=True)
plt.title('Boxplot of Trunk Height')
plt.ylabel('Trunk Height (meters)')

plt.subplot(2, 2, 2)
plt.boxplot(trees['Trunk_Diameter'], patch_artist=True)
plt.title('Boxplot of Trunk Diameter')
plt.ylabel('Trunk Diameter (meters)')

plt.subplot(2, 2, 3)
plt.boxplot(trees['Leaf_Length'], patch_artist=True)
plt.title('Boxplot of Leaf Length')
plt.ylabel('Leaf Length (cm)')

plt.subplot(2, 2, 4)
plt.boxplot(trees['Leaf_Width'], patch_artist=True)
plt.title('Boxplot of Leaf Width')
plt.ylabel('Leaf Width (cm)')


trees.describe()
```

**Trunk_Height**

```python
median_Trunk_Height = trees['Trunk_Height'].median() 
print(median_Trunk_Height)

percentile25 = trees['Trunk_Height'].quantile(0.25)
print('percentile25 ', percentile25)

percentile75 = trees['Trunk_Height'].quantile(0.75)
print('percentile75 ', percentile75)

iqr = percentile75 - percentile25
print(iqr)
```

```python
upper_limit = percentile75 + 1.5 *iqr
lower_limit = percentile25 - 1.5 * iqr

print ('Upper limit is: ' + str(upper_limit))
print ('Lower limit is: ' + str(lower_limit))
```

```python
df_lower = trees[trees['Trunk_Height'] < lower_limit]
df_lower
outliers_lower = df_lower['Trunk_Height'].tolist()
outliers_lower
trees['Trunk_Height'] = trees['Trunk_Height'].replace(outliers_lower, median_Trunk_Height)
```

```python
df_upper = trees[trees['Trunk_Height'] > upper_limit]
df_upper
outliers_upper = df_upper['Trunk_Height'].tolist()
outliers_upper
trees['Trunk_Height'] = trees['Trunk_Height'].replace(outliers_upper, median_Trunk_Height)
```

**Trunk_Diameter**	

```python
median_Trunk_Diameter = trees['Trunk_Diameter'].median() 
print(median_Trunk_Diameter)

percentile25 = trees['Trunk_Diameter'].quantile(0.25)
print('percentile25 ', percentile25)

percentile75 = trees['Trunk_Diameter'].quantile(0.75)
print('percentile75 ', percentile75)

iqr = percentile75 - percentile25
print(iqr)
```

```python
upper_limit = percentile75 + 1.5 *iqr
lower_limit = percentile25 - 1.5 * iqr

print ('Upper limit is: ' + str(upper_limit))
print ('Lower limit is: ' + str(lower_limit))
```

```python
df_lower = trees[trees['Trunk_Diameter'] < lower_limit]
df_lower
outliers_lower = df_lower['Trunk_Diameter'].tolist()
outliers_lower
trees['Trunk_Diameter'] = trees['Trunk_Diameter'].replace(outliers_lower, median_Trunk_Diameter)
```

```python
df_upper = trees[trees['Trunk_Diameter'] > upper_limit]
df_upper
outliers_upper = df_upper['Trunk_Diameter'].tolist()
outliers_upper
trees['Trunk_Diameter'] = trees['Trunk_Diameter'].replace(outliers_upper, median_Trunk_Diameter)
```

**Leaf_Width**

```python
median_Leaf_Width = trees['Leaf_Width'].median() 
print(median_Leaf_Width)

percentile25 = trees['Leaf_Width'].quantile(0.25)
print('percentile25 ', percentile25)

percentile75 = trees['Leaf_Width'].quantile(0.75)
print('percentile75 ', percentile75)

iqr = percentile75 - percentile25
print(iqr)
```

```python
upper_limit = percentile75 + 1.5 *iqr
lower_limit = percentile25 - 1.5 * iqr

print ('Upper limit is: ' + str(upper_limit))
print ('Lower limit is: ' + str(lower_limit))
```

```python
df_upper = trees[trees['Leaf_Width'] > upper_limit]
df_upper
outliers_upper = df_upper['Leaf_Width'].tolist()
outliers_upper
trees['Leaf_Width'] = trees['Leaf_Width'].replace(outliers_upper, median_Leaf_Width)
```

```python
plt.figure(figsize=(16, 8))


plt.subplot(2, 2, 1)
plt.boxplot(trees['Trunk_Height'], patch_artist=True)
plt.title('Boxplot of Trunk Height')
plt.ylabel('Trunk Height (meters)')

plt.subplot(2, 2, 2)
plt.boxplot(trees['Trunk_Diameter'], patch_artist=True)
plt.title('Boxplot of Trunk Diameter')
plt.ylabel('Trunk Diameter (meters)')

plt.subplot(2, 2, 3)
plt.boxplot(trees['Leaf_Length'], patch_artist=True)
plt.title('Boxplot of Leaf Length')
plt.ylabel('Leaf Length (cm)')

plt.subplot(2, 2, 4)
plt.boxplot(trees['Leaf_Width'], patch_artist=True)
plt.title('Boxplot of Leaf Width')
plt.ylabel('Leaf Width (cm)')


trees.describe()
```
