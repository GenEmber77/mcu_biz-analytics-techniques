 I will use the “Insurance in Sweden” sample dataset, which is part of the statsmodels package.

```python
import statsmodels.api as sm
import pandas as pd
import statsmodels.formula.api as smf
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python

# Load the "Insurance in Sweden" dataset
insurance_data = sm.datasets.get_rdataset("Insurance", package="MASS").data

# Display dataset information and structure
insurance_data.head()
```

Group refers to the engine capacity of the insured vehicle, measured in litres. <br>
<1l: less than 1 litre; 1–1.5l: between 1 and 1.5 litres etc <br>.

Age refers to the age group of the insurance policyholders, not the age of the vehicle. <25: younger than 25


```python
insurance_data.info()
```

We can see, that both Age and Group are of data type ‘object’, that’s because the two variables are categorical. We will have to keep that in mind for the regression.

For the purpose of this guide, we will analyze whether the Age of a person and the District they live in affect the number of insurance Claims they file – a count variable.

```python
# Let’s take a look at the distribution of claims by plotting it using a histogram.
histogram = insurance_data['Claims'].plot.hist()
```

As the histogram clearly shows , Claims features no negative values and it’s distribution is skewed. It certainly doesn’t look normally distributed.

# Poisson Regression

Generalized Linear Models (GLMs) allow us to model linear regressions for dependent variables of many different non-normal distributions. Analyzing count data specifically involves a systematic approach to model fitting, performance assessment, and handling complexities like overdispersion.

Let’s start by fittig a GLM based on the poisson distribution. To do so, I use the glm method from the statsmodels api. 

```python
# defining the poisson glm 
poisson_glm = smf.glm(formula = 'Claims ~ Age + District', data = insurance_data, family = sm.families.Poisson())

# fitting the model 
results = poisson_glm.fit()

# Printing the summary of the regression results
print(results.summary())
```

In the above code snippet, I model the effect of Age and District on the Number of Insurance Claims. Let’s take a look at the results:


There are three coefficients for Age, the fourth factor level was omitted and is used as a referece level. Keeping that in mind we can interpret the results as follows. Compared to people aged 20 to 29:

People aged 30–35 do not file significantly more or fewer insurance claims <br>
People younger than 25 file significantly fewer insurance claims <br>
People older than 35 file significantly more insurance claims <br>
Additionally, the lower the number of the district people live in, the less insurance claims they seem to file.

# Logistic regression

```python
df = pd.read_excel("dataset.xlsx")
df.head(2)
```

```python
print(df["transport_code"].unique())
print(df["transport_mode"].unique())
```

## Binary logistic regression
test if the age and income correlate with whethere the use buy the product again or not.

```python
X = df[['age', 'income']]
X = sm.add_constant(X)
y = df['purchased']  # Coded as 0 and 1

model = sm.Logit(y, X).fit()
print(model.summary())
```

# Multinomial logistic regression
test if age and income correlate with the type of transportaiton

```python

X = df[['age', 'income']]
X = sm.add_constant(X)
y = df['transport_code']  # For example: 0, 1 and 2

model = sm.MNLogit(y, X).fit()
print(model.summary())
```

```python
Transport code 0 is the reference category, or point of comparison. <br>
For example, the income coefficient of −2.264e−05 for transport code 1 means that as income increases, the likelihood of choosing transport category 1 rather than category 0 decreases, holding age constant.
```

# Ordinal logistic regression
Test if age and income correlate with satisfaction 

```python
from statsmodels.miscmodels.ordinal_model import OrderedModel
X = df[['age', 'income']]
y = df['satisfaction_code']  # For example: 1, 2 and 3
model = OrderedModel(
    y,
    X,
    distr='logit'
)

result = model.fit(method='bfgs')
print(result.summary())
```

# In the above result, the important inferential results are:

age: coefficient = 0.0263, p = 0.006 <br>
income: coefficient = 0.00003708, p < 0.001

Both coefficients are positive and statistically significant. Therefore, higher age and higher income are associated with a greater probability of being in a higher satisfaction category.

Age coefficient = 0.0263: Each additional year of age increases the odds of being in a higher satisfaction category by approximately 2.7%, holding income constant. <br>
Pseudo R²: how much better the fitted model performs compared with a model containing only the intercept.
