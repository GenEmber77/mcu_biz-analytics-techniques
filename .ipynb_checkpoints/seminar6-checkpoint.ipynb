Use the dataset below and run two regression analyses 

```python
import pandas as pd
#load dataset
wine = pd.read_csv("wine_sale.csv")

#examine first 5 rows of dataset
wine.head(5)
```

```python
wine.shape
```

```python
# Create pairwise scatterplots of Chinstrap penguins data
import seaborn as sns
sns.pairplot(wine)
```

linearity exists between profit and sales, and between rating and alcohol_content

```python
ols_data = wine[["Rating", "Alcohol_Content"]]
```

```python
# Write out formula  = Y~X
ols_formula = "Alcohol_Content ~ Rating"
```

```python
# Import ols function
from statsmodels.formula.api import ols
# Build OLS, fit model to data
OLS = ols(formula = ols_formula, data = ols_data)
model = OLS.fit()
model.summary()
```

# provide an explanation on OLS Regression Results. what is the conclusion?

It means higher ratings go with higher alcohol content. The relationship is very strong, and the numbers fit the pattern well. The rating score clearly matters, while the intercept doesn’t. The model’s checks show only small issues, so overall the results are trustworthy and the connection between rating and alcohol content is real and strong.
